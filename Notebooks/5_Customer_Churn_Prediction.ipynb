```python
import pandas as pd 
import numpy as np
import joblib
```

```python
model = joblib.load(
    r"C:\Users\Khande\Desktop\project 2\Customer-Churn-Prediction\models\random_forest_model.pkl"
)
```

```python
df = pd.read_csv(
    r"C:\Users\Khande\Desktop\project 2\Customer-Churn-Prediction\data\customer_churn_preprocessed.csv"
)
```

```python
sample_customer = df.drop("Churn", axis=1).iloc[[0]]

sample_customer
```

```python
prediction = model.predict(sample_customer)
prediction
```

```python
if prediction[0] == 1:
    print("The customer is likely to churn.")
else:
    print("The customer is not likely to churn.")
```

```python
probability = model.predict_proba(sample_customer)
print(probability)
```

```python
print(f"Probability of Staying : {probability[0][0]:.2f}%")
print(f"Probability of Churning : {probability[0][1]:.2f}%")
```

```python
customers = df.drop("Churn", axis=1).head(10)
predictions = model.predict(customers)
predictions
```

```python
results = customers.copy()
results["Prediction"] = predictions
results.head()
```

```python
results["Prediction"] = results["Prediction"].map({0: "Not Churn", 1: "Churn"})
results.head()
```

```python
results.to_csv(
    r"C:\Users\Khande\Desktop\project 2\Customer-Churn-Prediction\data\customer_predictions.csv",
    index=False
)
```
